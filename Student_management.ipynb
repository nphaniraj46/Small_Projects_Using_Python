```python
def add_student(): #i used this function to add a student to data
    name=input("Enter student name")
    age=int(input("Enter marks of the students"))#iam storing the data text file
    with open('student.txt','a') as f: #in this approach of handling a file we dont need to close this manually with this f.colse() Python will take care of that closing if we use this kind of approach
        f.write(f"{name},{age}\n")
    print("Student added succesfully")
def viwe_students(): #here in this function iam printing all the students data in the text file
    with open("student.txt","r") as f:
        print(f.readlines())#here we are going to get a list of elements
        f.seek(0)#iam resetting the cursour to start
        for i in f.readlines(): #printing the data line by line
            print(i)
def seach_student():
    name=input("Enter student name you want to seach for")
    with open("student.txt","r") as f:
        for i in f.readlines():
            i,m=i.strip().split(",") #here in this i we will get this name and in this m we will get marks because we are splitting the data into two right
            #here iam using strip to remove any back spaces or \n in the output
            if i==name:
                print("found")
                print(f"Name {name},Marks {m}")
            else:
                print(f"Student named {name} data not found in our data")
def main(): #here comes the main part this main part executes first
    while True:
        print("Select one option") #here we are selecting one option according to our needs
        print("1 ---> Add student")
        print("2 ---> viwe student")
        print("3 ---> search student")
        print("4 ---> Exit")
        
        choice=int(input("Enter you choice"))
        try:

            if choice==1:   #using exception hadling iam handling the  errors with appropriate messages that every one can understand what went wrong
                add_student()
            elif choice==2:
                viwe_students()
            elif choice==3:
                seach_student()
            elif choice==4:
                break
        except FileNotFoundError: 
            print("File is not found!") #This message is goin to be printed only when we get this FIleNotFOund error
        except Exception: #This message will be printed if we get any error other than this FileNotFound Error
            print("You might have entred or selected a wring input check out")
main()
```
