```python
S='Phoenix'
rev=''
print('Address before adding any elements',id(rev))
for i in S:
    rev=i+rev
    print(f'Address after adding {i} element',id(rev))

if rev==S:
    print('Pallindrome')
else:
    print('Not pallindrome')

#check output every time we are adding element to a string a new string is being created in a new address
#because strings are immutable

#same will be done for this tuple as well

#if we do like this then we will get high time complexity 0(n^2)

#in this place we need to use this two pointers method for solving this kind of problems
```

Pallindrome using this two pointer method which we have less time complexity

```python
s1=input('Enter a string')
left=0
right=len(s1)-1
is_pallindrome=True

while left<right:
    if s1[left]!=s1[right]:
        is_pallindrome=False
        break
    left+=1
    right-=1

if is_pallindrome:
    print(f'{s1} is Pallindrome')
else:
    print(f'{s1} is not a Pallindrome')
```

Practiced counting vowels spaces consonants digits special chars spaces

```python
s='Hello World 123!'

vowels = 0
consonants = 0
digits = 0
spaces = 0
special = 0
v='aeiou'
for i in s:  #here i used elif because if one condition is true i dont wanna check the other conditions
    if i in v:
        vowels+=1
    elif i not in v and i.isalpha():
        consonants+=1
    elif i==' ':
        spaces+=1
    elif i.isdigit():
        digits+=1
    else:
        special+=1

print(vowels)
print(consonants)
print(digits)
print(spaces)
print(special)
```
