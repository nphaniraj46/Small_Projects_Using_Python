```python
class Student:
    def __init__(self,name,age,branch):
        self.name=name
        self.age=age
        self.branch=branch
    def display(self):
        return {'Name':self.name,'Age':self.age,'Branch':self.branch}
    
student1=Student('Phani',22,'AI')
student2=Student('Rahul',22,'AIML')


s1_details=student1.display()
#print(sname,sage,sbranch)

s2_details=student2.display()
#print(sname,sage,sbranch)
print(s1_details)
print(s2_details)
```

```python
class Students1:
    def __init__(self,name):
        self.name=name
    def __str__(self):
        return self.name
roll_1=Students1('Phani')
print(roll_1)

#using this we are getting Phani as the output even when we are printing if from the referance variable roll_1
```

Encapsulation

```python
class BankAccount:
    def __init__(self,name,balance):
        self.name=name
        self.__balance=balance
    def deposite(self,amount):
        self.__balance+=amount
    def withdraw(self,amount):
        if amount>self.__balance:
            print('Insufficiant funds')
            return   #this stopes the next code of line incase this if condition is true
        self.__balance-=amount
    def showbalance(self):
        return self.__balance



    
        
```

```python
acc1=BankAccount('Phani',10)
```

```python
acc1.showbalance()
```

```python
acc1.withdraw(100)
```

```python
acc1.showbalance()
```
