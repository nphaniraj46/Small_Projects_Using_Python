```python
class Student:
    def __init__(self,name,age,branch):
        self.name=name
        self.age=age
        self.branch=branch
    def display(self):
        return {'Name':self.name,'Age':self.age,'Branch':self.branch}
    
student1=Student('Phani',22,'AI')
student2=Student('Rahul',22,'AIML')


s1_details=student1.display()
#print(sname,sage,sbranch)

s2_details=student2.display()
#print(sname,sage,sbranch)
print(s1_details)
print(s2_details)
```

```python
class Students1:
    def __init__(self,name):
        self.name=name
    def __str__(self):
        return self.name
roll_1=Students1('Phani')
print(roll_1)

#using this we are getting Phani as the output even when we are printing if from the referance variable roll_1
```
