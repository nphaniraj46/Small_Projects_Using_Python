```python
class BankAccount:
    def __init__(self,name,balance):
        self.name=name
        self.__balance=balance
    def deposite(self,amount):
        self.__balance+=amount
    def withdraw(self,amount):
        if amount>self.__balance:
            print('Insufficiant funds')
            return   #this stopes the next code of line incase this if condition is true
        self.__balance-=amount
    def showbalance(self):
        return self.__balance



    
        
```

```python
acc1=BankAccount('Phani',10)
```

```python
acc1.showbalance()
```

```python
def payment_process(payment_method,amount):
    payment_method.pay(amount)

class CC:
    def pay(self,amount):
        print(f'Paid {amount} through CreditCard')
class UPI:
    def pay(self,amount):
        print(f'Paid {amount} though UPI')
class Cash:
    def pay(self,amount):
        print(f'Paid {amount} though Cash')

payment_process(UPI(),100)
```

```python
from abc import ABC, abstractmethod

class Payment(ABC):

    @abstractmethod
    def pay(self):
        pass

class UPI(Payment):

    def pay(self):
        print("Payment using UPI")

class NetBanking(Payment):
    pass
```

```python
upi = UPI()
```

```python
upi.pay()
```

```python
bank = NetBanking()
bank.pay()
```
